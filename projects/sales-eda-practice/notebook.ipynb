```python
import matplotlib.pyplot as plt

# Step 1: Calculate the total sales grouped by Item Type
items = sales.groupby('Item Type')['Total Revenue'].sum()

# Step 2: Extract your separate index and values (Your logic)
itemsidx = items.index
itemsvals = items.values

# Step 3: Plot the bar chart
plt.figure(figsize=(10, 6)) # Makes the window wider for text space
plt.bar(itemsidx, itemsvals, color='skyblue', edgecolor='black')

# Step 4: Add titles and rotate text labels so they don't overlap
plt.title('Total Sales by Item Type')
plt.xlabel('Item Type')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45, ha='right') # Tilts long names like "Office Supplies"
plt.tight_layout() # Prevents text cutoff

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Step 1: Ensure 'Order Date' is in datetime format so it sorts chronologically
sales['Order Date'] = pd.to_datetime(sales['Order Date'])

# Step 2: Group by date and calculate order counts
# (Save to a fresh variable name like 'daily_sales' so you don't overwrite your dataframe!)
daily_sales = sales.groupby('Order Date')['Order ID'].count()

# Step 3: Extract separate indices (dates) and values (counts)
saleidx = daily_sales.index
saleval = daily_sales.values

# Step 4: Plot the line graph
plt.figure(figsize=(12, 5))  # Makes the window wider for timeline readability
plt.plot(saleidx, saleval, color='blue', linewidth=1.5, marker='o', markersize=3)

# Step 5: Format titles and labels
plt.title('Total Orders Over Time')
plt.xlabel('Order Date')
plt.ylabel('Number of Orders')
plt.grid(True, linestyle='--', alpha=0.6)  # Adds a clean background grid
plt.tight_layout()

plt.show()
```

```python

# Step 1: Isolate the two columns into separate variables
x_data = sales['Units Sold']
y_data = sales['Total Profit']

# Step 2: Create the figure and plot the points
plt.figure(figsize=(10, 6))
plt.scatter(x_data, y_data, color='purple', alpha=0.5, edgecolors='none', s=30)

# Step 3: Add formatting and labels
plt.title('Relationship Between Units Sold and Total Profit')
plt.xlabel('Units Sold')
plt.ylabel('Total Profit')
plt.grid(True, linestyle='--', alpha=0.5)  # Makes it easy to read coordinates
plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Loop through each numerical column and plot its distribution
for col in nums_cols:
    plt.figure(figsize=(6, 4))
    # Plots the histogram with a smooth density line over it
    sns.histplot(sales[col], kde=True, color='blue', bins=20)
    plt.title(f'Distribution Shape for {col}')
    plt.show()
```

```python
# Calculates correlation scores between all numeric columns
correlation_matrix = sales[nums_cols].corr()
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 8))

# Plot heatmap with scores rounded to 2 decimal places
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", vmin=-1, vmax=1)

plt.title('Correlation Matrix Heatmap')
plt.show()
```
